```python
import xgboost as xgb 
import pandas as pd 
import sklearn as sk 
import re 
import matplotlib.pyplot as plt
import seaborn as sns 
import numpy as np
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import KFold 
from sklearn.model_selection import GridSearchCV 
from sklearn.experimental import enable_halving_search_cv
from sklearn.model_selection import HalvingGridSearchCV
from sklearn.metrics import accuracy_score 
from scipy.linalg import sqrtm
from sklearn.utils import shuffle
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
%matplotlib inline
```

```python
def read_data(file,label):
    data = pd.read_csv(f'./train_data/select_data/{file}.csv')
    data['label'] = label
    return data
```

```python
a10 = read_data('m_a = 10 GeV',1)
a30 = read_data('m_a = 30 GeV',1)
a50 = read_data('m_a = 50 GeV',1)
a70 = read_data('m_a = 70 GeV',1)
a85 = read_data('m_a = 85 GeV',1)
atata = read_data('gammatautau',0)
```

```python
all_data = pd.concat([a10,a30,a50,a70,a85,atata],axis = 0)
all_data = all_data.sample(frac=1).reset_index(drop=True)
```

```python
a10.columns,all_data.shape
```

```python
X = all_data[['photon_energy','Photon_PT_0', 'deltar_mu_0_e_0', 'deltar_mu_0_gamma', 'deltar_e_0_gamma', 'MissingET_MET_0']]
y = all_data['label']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.5, random_state=42)
```

# 创建模型

```python
clf = xgb.XGBClassifier(learning_rate=0.02, 
                        max_depth=4, 
                        objective='binary:logistic',
                        eval_metric=['logloss', 'auc', 'error'],
                        tree_method='hist',  # 使用 CPU 的直方法
                        device='cuda',       # 指定使用 GPU
                        verbosity=2)
```

```python
model = xgb.XGBClassifier(learning_rate = 0.02 , max_depth=12, n_estimators=300,
                        objective='binary:logistic',
                        eval_metric=['logloss','auc','error'])
```

```python
model.fit(X_train,y_train)
```
